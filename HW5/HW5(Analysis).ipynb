1. Explain how implicit missing values impact medical datasets, and justify your choice of imputation strategy when handling these zeros compared to outright row deletion.
- Since implicit missing values represent unmeasured or corrupted data, but act as a true numerical value, they can have a large negative impact on the statistical distribution and the accuracy of a ML model. For my model, I ran into the issue of the value 0 being used as an unmeasured value in the Glucose, BloodPressure, SkinThickness, Insulin, and BMI columns. To handle these implicit values, I replaced all of the 0 measurements with NaN and used the SimpleImputer to replace them with the median of that column. I chose this over deleting the rows because I wanted to keep as much of the data as I could to work on my model and I wanted to avoid giving my model small amounts of data that could potentially bring a bias into question. Another area of confusion came from the pregnancies column, as the value 0 was used, but it was not clear whether that person truly did have zero pregnancies or that they weren't measured. Scientifically, the other columns cannot be zero.

2. Contrast how hard voting aggregates predictions versus soft voting. Why do well-calibrated probability estimates matter when implementing soft voting across heterogeneous models?
- Hard voting uses its predicted class labels from each model and chooses the class receiving the majority of support. Soft voting averages the predicted probabilities from each model and chooses the class with the highest average probability. Soft voting takes confidence into account for the model, while hard voting does not. This being said these probability must be well calibrated, or else large amounts of false confidence can push the model to make an inaccurate final prediction.
